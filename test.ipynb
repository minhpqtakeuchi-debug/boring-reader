````python
# pip install -U langchain langchain-google-genai google-generativeai pydantic
import os
import json
import re
from typing import Any, Dict

from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnableParallel, RunnablePassthrough

from app.core.config import settings


# ----------------------------
# CONFIG
# ----------------------------
API_KEY = settings.api_key

# Choose a Gemini model. For best quality/format reliability, use 1.5-pro at low temperature.
MODEL_NAME = settings.chatbot_model
TEMPERATURE = 0.0

# ----------------------------
# UTIL: load instruction text
# ----------------------------
def load_instruction_text(path: str) -> str:
    with open(path, "r", encoding="utf-8") as f:
        return f.read()

# ----------------------------
# UTIL: strip code fences and parse JSON
# ----------------------------
def extract_json(text: str) -> Dict[str, Any]:
    """
    Extract JSON from a string that may contain markdown fences.
    """
    # Grab content between ```json ... ``` or plain {...}
    fence_match = re.search(r"```json\s*(\{.*?\})\s*```", text, flags=re.S)
    if fence_match:
        payload = fence_match.group(1)
    else:
        # If no fenced block, try to find the first {...} JSON object
        brace_match = re.search(r"(\{.*\})", text, flags=re.S)
        if not brace_match:
            raise ValueError("No JSON object found in the model output.")
        payload = brace_match.group(1)

    # Parse
    return json.loads(payload)

# ----------------------------
# CORE: build the chain
# ----------------------------
# ----------------------------
# CORE: build the chain  (FIXED)
# ----------------------------
def build_chain(instruction_text: str):
    """
    Build a simple LangChain pipeline:
      system: instruction prompt (passed as a variable so { } aren't parsed)
      user:   raw boring JSON
    Returns a Runnable that takes {"boring_json": <dict or str>} and outputs parsed dict.
    """
    system_part = (
        instruction_text.strip()
        + "\n\n"
        + "**Important:** Return only the final JSON object per the schema."
    )

    # IMPORTANT: pass instruction as a template variable, not inline literal
    prompt = ChatPromptTemplate.from_messages(
        [
            ("system", "{instruction}"),
            (
                "user",
                "Here is the boring JSON (parsed from PDF). Use ONLY this content:\n\n{boring_json}\n",
            ),
        ]
    )

    llm = ChatGoogleGenerativeAI(
        model=MODEL_NAME,
        temperature=TEMPERATURE,
        api_key=API_KEY,
        convert_system_message_to_human=True,  # Gemini shim
    )

    # Runnable pipeline:
    chain = (
        {
            "boring_json": RunnablePassthrough(),               # will be filled at invoke time
            "instruction": lambda x: system_part,               # inject the raw instruction text as-is
        }
        | prompt
        | llm
        | (lambda msg: msg.content)  # extract text
        | extract_json               # parse JSON into dict
    )

    return chain

# ----------------------------
# PUBLIC: transform function  (minor tweak: nothing to change except we already handle)
# ----------------------------
def transform_boring_json(
    boring_json: Any,
    instruction_path: str = "boring_log_instructions.txt",
) -> Dict[str, Any]:
    """
    - boring_json: dict or JSON string of your 'boring' file content.
    - instruction_path: path to the .txt instruction you saved earlier.
    Returns: normalized JSON (as a Python dict) per your schema.
    """
    if isinstance(boring_json, dict):
        boring_json_str = json.dumps(boring_json, ensure_ascii=False, indent=2)
    else:
        boring_json_str = str(boring_json)

    instruction_text = load_instruction_text(instruction_path)
    chain = build_chain(instruction_text)

    # Because we used RunnablePassthrough for 'boring_json' and a lambda for 'instruction',
    # we only need to pass the boring_json_str when invoking:
    result = chain.invoke(boring_json_str)
    return result
````

````python
import json
import re
from typing import Any, Dict, Optional

from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnablePassthrough

from app.core.config import settings


class BoringJSONAgent:
    @staticmethod
    def _load_instruction_text(path: str) -> str:
        with open(path, "r", encoding="utf-8") as f:
            return f.read()

    @staticmethod
    def _extract_json(text: str) -> Dict[str, Any]:
        fence = re.search(r"```json\s*(\{.*?\})\s*```", text, flags=re.S)
        if fence:
            payload = fence.group(1)
        else:
            brace = re.search(r"(\{.*\})", text, flags=re.S)
            if not brace:
                raise ValueError("No JSON object found in the model output.")
            payload = brace.group(1)
        return json.loads(payload)

    def __init__(
        self,
        instruction_path: str,
        api_key: Optional[str] = None,
        model_name: Optional[str] = None,
        temperature: float = 0.0,
        json_only: bool = True,
        max_retries: int = 1,
    ):
        self.instruction_path = instruction_path
        self.api_key = api_key or settings.api_key
        self.model_name = model_name or settings.chatbot_model
        self.temperature = temperature
        self.json_only = json_only
        self.max_retries = max_retries

        self._instruction_text = self._load_instruction_text(self.instruction_path)
        self._chain = self._build_chain(self._instruction_text)

    def _build_chain(self, instruction_text: str):
        system_part = instruction_text.strip() + (
            "\n\nIMPORTANT:\n"
            "- Return only one JSON object.\n"
            "- No explanations. No code fences. No markdown.\n"
        )

        # Build prompt and permanently "partial" the instruction into it
        prompt = ChatPromptTemplate.from_messages(
            [
                ("system", "{instruction}"),
                (
                    "user",
                    "Here is the boring JSON (parsed from PDF). Use ONLY this content:\n\n{boring_json}\n",
                ),
            ]
        ).partial(instruction=system_part)  # ← baked in once

        llm_kwargs = {}
        if self.json_only:
            llm_kwargs["response_mime_type"] = "application/json"

        llm = ChatGoogleGenerativeAI(
            model=self.model_name,
            temperature=self.temperature,
            api_key=self.api_key,
            convert_system_message_to_human=True,
            model_kwargs=llm_kwargs,
            max_retries=self.max_retries,
        )

        chain = (
            {"boring_json": RunnablePassthrough()}  # only var left per-call
            | prompt
            | llm
            | (lambda msg: msg.content)
            | self._extract_json
        )
        return chain

    def convert(self, boring_json: Any) -> Dict[str, Any]:
        # Compact JSON to reduce tokens
        if isinstance(boring_json, dict):
            payload = json.dumps(boring_json, ensure_ascii=False, separators=(",", ":"))
        else:
            payload = str(boring_json)
        return self._chain.invoke(payload)

    def reload_instruction(self):
        self._instruction_text = self._load_instruction_text(self.instruction_path)
        self._chain = self._build_chain(self._instruction_text)

    def update_model(self, model_name: Optional[str] = None, temperature: Optional[float] = None):
        if model_name:
            self.model_name = model_name
        if temperature is not None:
            self.temperature = temperature
        self._chain = self._build_chain(self._instruction_text)
````

````python
# Build once; instruction stays in memory
agent = BoringJSONAgent(
    instruction_path=settings.chatbot_instruction,  # e.g., "boring_log_instructions.txt"
    api_key=settings.api_key,
    model_name=settings.chatbot_model,              # e.g., "gemini-1.5-flash" for speed
    temperature=0.0,
)
````

````python
# Example boring JSON (dict or string)
example_boring = json_page0

normalized = agent.convert(example_boring)
print(json.dumps(normalized, ensure_ascii=False, indent=2))
````
